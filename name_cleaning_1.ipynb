```python
import pandas as pd
```

```python
train_source1 = pd.read_csv(
    "train_source1.tsv",
    sep="\t"
)

train_source2 = pd.read_csv(
    "train_source2.tsv",
    sep="\t"
)

train_source3 = pd.read_csv(
    "train_source3.tsv",
    sep="\t"
)

train_ground_truth = pd.read_csv(
    "train_ground_truth.tsv",
    sep="\t"
)
```

```python
import re
import unicodedata
import pandas as pd

def clean_business_name(text):
    # 1. Handle missing values
    if pd.isna(text):
        return ""

    # 2. Convert to string
    text = str(text)

    # 3. Unicode normalization
    text = unicodedata.normalize("NFKC", text)

    # 4. Lowercase
    text = text.lower()

    # 5. Remove apostrophes without adding a space
    text = text.replace("'", "")

    # 6. Remove domain suffixes
    # Examples:
    # abc.com     -> abc
    # abc.in      -> abc
    # abc.co.in   -> abc
    # abc.org     -> abc
    # abc.net     -> abc
    text = re.sub(
        r"\.(?:com|in|co\.in|org|net|biz|info|edu|gov)$",
        "",
        text
    )

    # 7. Remove remaining punctuation
    text = re.sub(r"[^\w\s]", " ", text)

    # 8. Normalize whitespace
    text = re.sub(r"\s+", " ", text).strip()

    return text
```

```python
train_source1["business_name_clean"] = train_source1["business_name"].apply(clean_business_name)

train_source2["business_name_clean"] = train_source2["business_name"].apply(clean_business_name)

train_source3["business_name_clean"] = train_source3["business_name"].apply(clean_business_name)
```
