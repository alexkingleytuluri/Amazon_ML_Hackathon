# 3. ADDRESS CLEANING
###  Member 3

This section normalizes business addresses across Source 1, Source 2, and Source 3.

The original address columns are preserved.
Cleaned addresses are stored in separate `clean_address` columns.

```python
# ==========================================
# ADDRESS DATA INSPECTION
# ==========================================

print("SOURCE 1 ADDRESS EXAMPLES")
display(
    train_source1[["business_address", "country"]]
    .sample(20, random_state=42)
)

print("\nMISSING ADDRESS COUNT")
print(train_source1["business_address"].isna().sum())

print("\nTOTAL ADDRESSES")
print(len(train_source1))
```

```python
# ==========================================
# ADDRESS CLEANING FUNCTION
# ==========================================

import re

def clean_address(address):
    """
    Normalize a business address for matching.
    Original address is not modified.
    """

    # Handle missing values
    if pd.isna(address):
        return ""

    # Convert to string
    address = str(address)

    # Lowercase
    address = address.lower()

    # Replace common punctuation/separators with spaces
    address = re.sub(r"[,./#\-]", " ", address)

    # Common address abbreviations
    replacements = {
        r"\brd\b": "road",
        r"\bst\b": "street",
        r"\bave\b": "avenue",
        r"\bblvd\b": "boulevard",
        r"\bln\b": "lane",
        r"\bdr\b": "drive",
        r"\bct\b": "court",
        r"\bhwy\b": "highway",
    }

    for pattern, replacement in replacements.items():
        address = re.sub(pattern, replacement, address)

    # Normalize multiple spaces
    address = re.sub(r"\s+", " ", address).strip()

    return address
```

```python
# ==========================================
# TEST ADDRESS CLEANER
# ==========================================

test_addresses = [
    "1795 Westchester Drive, High Point, NC",
    "105 ELM ST, MORGANTON, NC",
    "1712 Montebello Avenue, Phoenix, AZ",
    "KH NO. -570/13, NEW DELHI, WEST DELHI, Delhi",
    None
]

for address in test_addresses:
    print("ORIGINAL:", address)
    print("CLEANED :", clean_address(address))
    print("-" * 70)
```

```python
# ==========================================
# CREATE CLEAN ADDRESS COLUMN
# ==========================================

train_source1["clean_address"] = train_source1["business_address"].apply(clean_address)

print("Address cleaning completed!")
print(train_source1.columns.tolist())
```

```python
# ==========================================
# APPLY ADDRESS CLEANING TO SOURCE 2 & 3
# ==========================================

train_source2["clean_address"] = (
    train_source2["business_address"].apply(clean_address)
)

train_source3["clean_address"] = (
    train_source3["business_address"].apply(clean_address)
)

print("✅ Address cleaning completed for Source 2 and Source 3.")
```

## Address Cleaning — Member 3 Complete

- Inspected Source 1 addresses.
- Source 1 contains 2,206,821 addresses with no missing values.
- Created `clean_address()` normalization function.
- Preserved original `business_address`.
- Created `clean_address` as a separate column.
- Cleaner handles lowercase conversion, punctuation normalization, common address abbreviations, and whitespace normalization.
- The same function has been applied to Source 1, Source 2, and Source 3.

**Name Cleaning Member 2 Work**

```python
import re
import unicodedata
import pandas as pd

def clean_business_name(text):
    # 1. Handle missing values
    if pd.isna(text):
        return ""

    # 2. Convert to string
    text = str(text)

    # 3. Unicode normalization
    text = unicodedata.normalize("NFKC", text)

    # 4. Lowercase
    text = text.lower()

    # 5. Remove apostrophes without adding a space
    text = text.replace("'", "")

    # 6. Remove domain suffixes
    # Examples:
    # abc.com     -> abc
    # abc.in      -> abc
    # abc.co.in   -> abc
    # abc.org     -> abc
    # abc.net     -> abc
    text = re.sub(
        r"\.(?:com|in|co\.in|org|net|biz|info|edu|gov)$",
        "",
        text
    )

    # 7. Remove remaining punctuation
    text = re.sub(r"[^\w\s]", " ", text)

    # 8. Normalize whitespace
    text = re.sub(r"\s+", " ", text).strip()

    return text
```

```python
# ==========================================
# APPLY NAME CLEANING TO ALL TRAINING SOURCES
# ==========================================

train_source1["business_name_clean"] = (
    train_source1["business_name"].apply(clean_business_name)
)

train_source2["business_name_clean"] = (
    train_source2["business_name"].apply(clean_business_name)
)

train_source3["business_name_clean"] = (
    train_source3["business_name"].apply(clean_business_name)
)

print("✅ Business name cleaning completed for Source 1, Source 2 and Source 3.")
```

```python
# ==========================================
# VERIFY NAME CLEANING
# ==========================================

print("Source 1:")
display(
    train_source1[
        ["business_name", "business_name_clean"]
    ].sample(10, random_state=42)
)

print("\nSource 2:")
display(
    train_source2[
        ["business_name", "business_name_clean"]
    ].sample(10, random_state=42)
)

print("\nSource 3:")
display(
    train_source3[
        ["business_name", "business_name_clean"]
    ].sample(10, random_state=42)
)
```
